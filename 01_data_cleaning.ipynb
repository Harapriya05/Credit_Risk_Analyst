```python
df['account_open_date']=pd.to_datetime(df['account_open_date'])
```

```python
#lamda x function is used for check every value of x =days_past_due values
df['risk_flag']=df['days_past_due'].apply(lambda x: 'bad' if x>60 else 'good')
```

```python
#Categorizes customers into:High risk,Medium risk,Low risk
def risk_catagory(x):
    if x>60:
        return "High"
    elif x>30:
        return "medium"
    else:
        return "low"
df['risk_catagory']=df['days_past_due'].apply(risk_catagory)
```

```python
# Calculates how long customer is with bank
df['vintage_days']=(pd.to_datetime('today')-df['account_open_date']).dt.days 
```

```python
#if balance is 500 and limit=1000 So, 500/1000=0.5=50% (utilization=50%)
#High utilization → high risk
df['utilization']=df['balance']/df['credit_limit']
```
