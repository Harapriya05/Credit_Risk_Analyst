```python
df['utilization']=df['balance']/df['credit_limit'] #Calculates % of credit used
```

```python
charged_off=df[df['status']=='charged_off'] ['balance'].sum()
total_balance = df['balance'].sum()

recovery_rate = charged_off / total_balance
```

```python
def collection_strategy(x):
    if x == 'High':
        return 'Aggressive'
    elif x == 'Medium':
        return 'Moderate'
    else:
        return 'Soft'

df['collection_strategy'] = df['risk_level'].apply(collection_strategy) #Assigns strategy to each customer
```

```python
df.sort_values(by='balance', ascending=False).head() #Shows customers with highest balance,“Which customers can cause biggest loss?”
```
